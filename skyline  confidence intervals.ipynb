```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("skyline_enrollments.csv")
x = df["hours_studied"].dropna()
x.describe()
```

```python
n     = len(x)
mean  = x.mean()
sd    = x.std(ddof=1)                 # sample standard deviation
se    = sd / np.sqrt(n)               # standard error
tcrit = stats.t.ppf(0.975, df=n-1)    # t critical value, 95%
moe   = tcrit * se                    # margin of error
ci    = (mean - moe, mean + moe)

print(f"Sample size (n):      {n}")
print(f"Sample mean:          {mean:.4f}")
print(f"Sample std dev (s):   {sd:.4f}")
print(f"Standard error:       {se:.4f}")
print(f"t critical (df={n-1}): {tcrit:.4f}")
print(f"Margin of error:      {moe:.4f}")
print(f"95% CI: ({ci[0]:.4f}, {ci[1]:.4f})")
```
